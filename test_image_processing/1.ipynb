```python
import cv2
import numpy as np

height, width = 1200, 1800

bg_color = (95, 145, 105) 
cross_color = (100, 200, 220) 

image = np.full((height, width, 3), bg_color, dtype=np.uint8)

vertical_rect_top_left = (int(width * 0.4), int(height * 0.2))      # (x1, y1)
vertical_rect_bottom_right = (int(width * 0.6), int(height * 0.8))  # (x2, y2)
cv2.rectangle(image, vertical_rect_top_left, vertical_rect_bottom_right, cross_color, -1)

horizontal_rect_top_left = (int(width * 0.2), int(height * 0.4))    # (x1, y1)
horizontal_rect_bottom_right = (int(width * 0.8), int(height * 0.6))# (x2, y2)
cv2.rectangle(image, horizontal_rect_top_left, horizontal_rect_bottom_right, cross_color, -1)

cv2.imshow("Flag Image", image)
cv2.imwrite("flag.png", image)
flag = cv2.imread('flag.png')
flag.shape
flag_bgr=cv2.cvtColor(flag,cv2.COLOR_BGR2RGB)
plt.imshow(flag_bgr)
```

```python
tiles = {}
count = 1
H_tile, W_tile = 300, 450

for r in range(4):
    for c in range(4):
        # Xác định tọa độ vùng cắt
        y1, y2 = r * H_tile, (r + 1) * H_tile
        x1, x2 = c * W_tile, (c + 1) * W_tile
        
        # Cắt mảnh ảnh và lưu vào dictionary với khóa là số thứ tự (1..16)
        tiles[count] = flag[y1:y2, x1:x2].copy()
        count += 1
```

```python
grid_order = [
    [12,  1,  5,  7],
    [16,  2,  9,  8],
    [ 3,  6,  4, 10],
    [11, 14, 15, 13]
]
# Ghép theo từng hàng rồi xếp chồng lên nhau
rows = []
for row_indices in grid_order:
    row_imgs = [tiles[idx] for idx in row_indices]
    rows.append(np.hstack(row_imgs))
flag_scrambled = np.vstack(rows)
# Lưu ảnh bị xáo trộn ra file .jpg
cv2.imwrite("scrambled_flag.jpg", flag_scrambled)
flag_scrambled_bgr=cv2.cvtColor(flag_scrambled,cv2.COLOR_BGR2RGB)
plt.imshow(flag_scrambled_bgr)
```

```python
import os
import cv2
import matplotlib.pyplot as plt

# 1. Đọc ảnh gốc (Ảnh 1)
img1 = cv2.imread("shin.jpg")
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

# 2. Tạo ảnh thứ 2 (Ví dụ: Ảnh lượng tử hóa 64 mức)
step = 256 // 64  # step = 4
img2_rgb = (img1_rgb // step) * step

# Lưu thử 2 ảnh ra đĩa để lấy dung lượng file (tùy chọn)
cv2.imwrite("img1_temp.png", cv2.cvtColor(img1_rgb, cv2.COLOR_RGB2BGR))
cv2.imwrite("img2_temp.png", cv2.cvtColor(img2_rgb, cv2.COLOR_RGB2BGR))

# 3. Tính dung lượng RAM & Dung lượng File của từng ảnh
# --- Ảnh 1 ---
ram1_kb = img1_rgb.nbytes / 1024  # Dung lượng RAM (KB)
file1_kb = os.path.getsize("img1_temp.png") / 1024  # Dung lượng File đĩa (KB)

# --- Ảnh 2 ---
ram2_kb = img2_rgb.nbytes / 1024  # Dung lượng RAM (KB)
file2_kb = os.path.getsize("img2_temp.png") / 1024  # Dung lượng File đĩa (KB)

# 4. In thông tin ra màn hình Chat / Console
print(
    f"Ảnh 1: RAM = {ram1_kb:.2f} KB | File = {file1_kb:.2f} KB | Shape = {img1_rgb.shape}"
)
print(
    f"Ảnh 2: RAM = {ram2_kb:.2f} KB | File = {file2_kb:.2f} KB | Shape = {img2_rgb.shape}"
)

# 5. Hiển thị 2 ảnh lên đồ thị Matplotlib kèm dung lượng trên Title
plt.figure(figsize=(12, 6))

# Hiển thị Ảnh 1
plt.subplot(1, 2, 1)
plt.imshow(img1_rgb)
plt.title(
    f"Ảnh gốc (256 mức)\nRAM: {ram1_kb:.1f} KB | File: {file1_kb:.1f} KB",
    fontsize=12,
    color="blue",
)
plt.axis("off")

# Hiển thị Ảnh 2
plt.subplot(1, 2, 2)
plt.imshow(img2_rgb)
plt.title(
    f"Ảnh 64 mức (6-bit)\nRAM: {ram2_kb:.1f} KB | File: {file2_kb:.1f} KB",
    fontsize=12,
    color="red",
)
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
def contrast_stretching(img, r1, s1, r2, s2, L=256):
    r = img.astype(np.float64)
    s = np.zeros_like(r)
    # Đoạn 1: từ 0 đến r1
    mask1 = r < r1
    s[mask1] = (s1 / r1) * r[mask1]
    # Đoạn 2: từ r1 đến r2
    mask2 = (r >= r1) & (r <= r2)
    s[mask2] = s1 + ((s2 - s1) / (r2 - r1)) * (r[mask2] - r1)
    # Đoạn 3: từ r2 đến L-1 (255)
    mask3 = r > r2
    s[mask3] = s2 + (((L - 1) - s2) / ((L - 1) - r2)) * (r[mask3] - r2)
    return np.clip(s, 0, L - 1).astype(np.uint8)

# Chọn 2 điểm mốc tương ứng với hình vẽ trong slide
r1, s1 = 90, 30
r2, s2 = 160, 220
# Thực hiện biến đổi
img_contrast = contrast_stretching(img_gray, r1, s1, r2, s2)
# --- VẼ ĐỒ THỊ VÀ ẢNH SO SÁNH ---
r_vals = np.arange(256)
T_vals = contrast_stretching(r_vals, r1, s1, r2, s2)
plt.figure(figsize=(14, 5))
# 1. Đồ thị T(r)
plt.subplot(1, 3, 1)
plt.plot(r_vals, T_vals, "r-", linewidth=2.5, label="Hàm T(r)")
plt.plot([r1, r2], [s1, s2], "ro")  # Chấm 2 điểm mốc
plt.text(r1 - 15, s1 + 10, f"({r1},{s1})", fontsize=10)
plt.text(r2 - 25, s2 - 15, f"({r2},{s2})", fontsize=10)
plt.title("Đồ thị Contrast Stretching")
plt.xlabel("r (Input)")
plt.ylabel("s (Output)")
plt.grid(True, linestyle=":")
plt.legend()
# 2. Ảnh gốc
plt.subplot(1, 3, 2)
plt.imshow(img_gray, cmap="gray")
plt.title("Ảnh gốc (Độ tương phản thấp)")
plt.axis("off")
# 3. Ảnh kết quả
plt.subplot(1, 3, 3)
plt.imshow(img_contrast, cmap="gray")
plt.title("Ảnh sau Contrast Stretching")
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# ==========================================================
# Biến đổi cắt ngưỡng mức xám (Intensity-Level Slicing) - Figure 3.11
# ==========================================================
import numpy as np

def intensity_slicing_a(img_gray, A, B, L_high=255, L_low=20):
    """
    (a) Highlight dải độ sáng [A, B] và giảm các độ sáng khác xuống mức thấp (L_low).
    """
    res = np.full_like(img_gray, L_low)
    mask = (img_gray >= A) & (img_gray <= B)
    res[mask] = L_high
    return res

def intensity_slicing_b(img_gray, A, B, L_high=255):
    """
    (b) Highlight dải độ sáng [A, B] và giữ nguyên các mức độ sáng khác.
    """
    res = img_gray.copy()
    mask = (img_gray >= A) & (img_gray <= B)
    res[mask] = L_high
    return res

# Thiết lập ngưỡng A và B (có thể điều chỉnh tùy thuộc vào ảnh của bạn)
A, B = 100, 180
img_sliced_a = intensity_slicing_a(img_gray, A, B)
img_sliced_b = intensity_slicing_b(img_gray, A, B)

# Hiển thị kết quả so sánh 3 ảnh
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.title("Ảnh xám gốc")
plt.imshow(img_gray, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title(f"Transformation (a)\n[A={A}, B={B}] (Giảm nền)")
plt.imshow(img_sliced_a, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title(f"Transformation (b)\n[A={A}, B={B}] (Giữ nền)")
plt.imshow(img_sliced_b, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
df_orig = pd.read_csv('data.csv')

df_orig['Series'] = df_orig['Series'].astype(str).str.split().str[0]

frame_keywords = [
    'FRAME', 'ISOMETRIC', 'SWORD', 'BOX', 'DIAMOND', 'HEX', 'OCTABLADE',
    'FORMATION', 'COMPACT', 'RAZOR', 'THIN FRAME', 'GROMMET', 'BUMPER',
    'POWER RING', 'ENERGY BOW', 'SONIC FLARE', 'TRI-VOLTAGE', 'VOLTAGE',
    'CATAPULT', 'POCKETING'
]
material_keywords = [
    'NANO', 'CARBON', 'FIBER', 'TUNGSTEN', 'PYROFIL', 'HMG', 'M40X',
    'TORAYCA', 'FULLERENE', 'METALLIC', 'PBO', 'GRAPHITE', 'NAMD',
    'REXIS', 'NANOCELL', 'NEO CS'
]
stability_keywords = [
    'TORSION', 'JOINT', 'T-JOINT', 'ANCHOR', 'SHAFT', 'CORE', 'SYSTEM',
    'GENERATOR', 'ROTATIONAL', 'BOOST', 'CONTROL', 'VIBRATION', 'DAMPEN',
    'STABILIT', 'WES', 'CSR', 'CFR', 'FREECORE', 'FREE CORE', 'FREE-CORE',
    'REBOUND', 'WHIPPING', 'CAP', 'FILTER', 'SOUND'
]

def get_group_names(tech_str, keywords):
    if pd.isna(tech_str):
        return ''
    techs = [t.strip() for t in str(tech_str).split(',')]
    matched = [t for t in techs if any(kw.upper() in t.upper() for kw in keywords)]
    return ', '.join(matched) if matched else ''

df_orig['Tech_Frame'] = df_orig['Technologies'].apply(lambda x: get_group_names(x, frame_keywords))
df_orig['Tech_Material'] = df_orig['Technologies'].apply(lambda x: get_group_names(x, material_keywords))
df_orig['Tech_Stability'] = df_orig['Technologies'].apply(lambda x: get_group_names(x, stability_keywords))

df_orig['Version'] = df_orig['Price'].apply(lambda x: 'v3' if x > 3_000_000 else ('v2' if x >= 1_500_000 else 'v1'))

df_orig = df_orig.drop(columns=['Technologies'])
df_orig = df_orig.sort_values(['Brand', 'Series', 'Price']).reset_index(drop=True)

df_orig.to_csv('dataprocessing.csv', index=False)
print("Đã lưu!")
df_orig.head(10)
```

```python
import pandas as pd
import numpy as np
```

```python
path_shopvnb = r"d:\dut_ai\AIO_code\ProjectTabular\data\badminton_rackets.csv"
path_hvshop = r"d:\dut_ai\AIO_code\ProjectTabular\data\badminton_hvshop.csv"
path_fbshop = r"d:\dut_ai\AIO_code\ProjectTabular\data\data_fbshop.csv"
path_bbay = r"d:\dut_ai\AIO_code\ProjectTabular\data\badmintonbay.csv"
path_r4u = r"d:\dut_ai\AIO_code\ProjectTabular\data\racquets4u_all_brands.csv"

df_shopvnb = pd.read_csv(path_shopvnb)
df_hvshop = pd.read_csv(path_hvshop)
df_fbshop = pd.read_csv(path_fbshop)
df_bbay = pd.read_csv(path_bbay)
df_r4u = pd.read_csv(path_r4u)

df_shopvnb['Source'] = 'ShopVNB'
df_hvshop['Source'] = 'HVShop'
df_fbshop['Source'] = 'FBShop'
df_bbay['Source'] = 'BadmintonBay'
df_r4u['Source'] = 'Racquets4U'

print(df_shopvnb.columns)
print(df_hvshop.columns)
print(df_fbshop.columns)
print(df_bbay.columns)
```

```python
import re
import numpy as np

# 1. Hàm làm sạch và chuyển đổi giá
def clean_price(row):
    price_str = str(row['Price'])
    source = row['Source']
    
    if pd.isna(row['Price']) or price_str.strip() == '' or price_str.lower() in ['nan', 'none']:
        return np.nan
        
    if source == 'BadmintonBay':
        try:
            num_str = re.sub(r'[^\d.]', '', price_str)
            usd_price = float(num_str)
            return int(usd_price * 25000)
        except:
            return np.nan

    elif source == 'Racquets4U':
        try:
            # Price in Indian Rupees, remove non-digits
            num_str = re.sub(r'[^\d]', '', price_str)
            inr_price = float(num_str)
            # Roughly 1 INR = 300 VND
            return int(inr_price * 300)
        except:
            return np.nan
    else:
        try:
            num_str = re.sub(r'[^\d]', '', price_str)
            return int(num_str)
        except:
            return np.nan

df_final['Price_VND'] = df_final.apply(clean_price, axis=1)

# 2. Hàm chuẩn hoá Trọng Lượng (Weight)
def extract_weight_U(text):
    text = str(text).upper()
    match = re.search(r'([1-6]U)', text)
    if match:
        return match.group(1)
    
    match_g = re.search(r'(\d{2})[-~]?\s*g', text, re.IGNORECASE)
    if match_g:
        weight = int(match_g.group(1))
        if weight >= 90: return '2U'
        elif 85 <= weight <= 89: return '3U'
        elif 80 <= weight <= 84: return '4U'
        elif 75 <= weight <= 79: return '5U'
        elif 70 <= weight <= 74: return '6U'
        
    return np.nan

df_final['Weight_Class'] = df_final['Weight'].apply(extract_weight_U)

# 3. Chuẩn hoá Điểm Cân Bằng (Balance point)
def clean_balance_point(text):
    text = str(text).lower()
    match_mm = re.search(r'(\d{3})', text)
    if match_mm:
        val = int(match_mm.group(1))
        if 270 <= val <= 330:
            return val
            
    if 'heavy' in text or 'nặng đầu' in text:
        return 300
    elif 'even' in text or 'cân bằng' in text:
        return 290
    elif 'light' in text or 'nhẹ đầu' in text:
        return 285
        
    return np.nan

df_final['Balance_Point_mm'] = df_final['Balance'].apply(clean_balance_point)

# 4. Chuẩn Hoá Tên Hãng (Brand)
df_final['Brand'] = df_final['Brand'].astype(str).str.strip().str.upper()
df_final['Brand'] = df_final['Brand'].replace('NAN', np.nan)

display(df_final[['Source', 'Brand', 'Price', 'Price_VND', 'Weight', 'Weight_Class', 'Balance', 'Balance_Point_mm']].head(15))

```

```python
# Xoá các cột gốc chưa làm sạch để dữ liệu gọn gàng hơn
df_final = df_final.drop(columns=['Price', 'Weight', 'Balance'])

# Đổi tên các cột đã làm sạch lại thành tên gốc (nếu bạn muốn)
df_final = df_final.rename(columns={
    'Price_VND': 'Price',
    'Weight_Class': 'Weight',
    'Balance_Point_mm': 'Balance'
})

display(df_final.head())

# Lưu lại file csv đã sạch hoàn toàn
df_final.to_csv(r"d:\dut_ai\AIO_code\ProjectTabular\data\data_merged_all.csv", index=False, encoding='utf-8-sig')
```

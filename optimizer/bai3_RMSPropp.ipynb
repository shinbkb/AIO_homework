```python
import numpy as np
```

```python
def df_w(W):
    w1,w2 = W[0],W[1]
    dw1 = 0.2 * w1
    dw2 = 4*w2
    return np.array([dw1,dw2],dtype=np.float32)
```

```python
def rmsprop(S,W,dw,lr,gamma,epsilon=1e-6):
    S=gamma * S +(1-gamma)*(dw**2)
    W=W - lr * (dw/(np.sqrt(S+epsilon)))
    return W,S
```

```python
def train_p1_rmsprop(lr, gamma, epochs):
    # Khởi tạo điểm bắt đầu và biến S
    W = np.array([-5, -2], dtype=np.float32)
    S = np.array([0, 0], dtype=np.float32)
    
    results = [W.copy()] # Lưu lịch sử (dùng .copy() để tránh ghi đè)
    
    for i in range(epochs):
        # Tìm gradient
        dw = df_w(W)
        
        # Cập nhật theo RMSProp
        W, S = rmsprop(W, S, dw, lr, gamma)
        
        # Lưu kết quả sau mỗi epoch
        results.append(W.copy())
        
    return results
```

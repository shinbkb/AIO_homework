```python
import numpy as np
```

```python
def df_w(W):
    w1,w2 = W[0],W[1]
    dw1 = 0.2 * w1
    dw2 = 4*w2
    return np.array([dw1,dw2],dtype=np.float32)
```

```python
def adam_optimizer(W, V, S, dw, lr, beta1, beta2, t, epsilon=1e-6):
    V = beta1 * V + (1 - beta1) * dw
    
    S = beta2 * S + (1 - beta2) * (dw**2)
    
    V_corr = V / (1 - beta1**t)
    S_corr = S / (1 - beta2**t)
    
    W = W - lr * V_corr / (np.sqrt(S_corr) + epsilon)
    
    return W, V, S
```

```python
def train_p1_adam(lr, beta1, beta2, epochs):
    # Khởi tạo các giá trị ban đầu [cite: 246]
    W = np.array([-5, -2], dtype=np.float32)
    V = np.array([0, 0], dtype=np.float32)
    S = np.array([0, 0], dtype=np.float32)
    
    results = [W.copy()]
    
    for t in range(1, epochs + 1): # t bắt đầu từ 1 cho bias correction
        # Tìm gradient
        dw = df_w(W)
        
        # Cập nhật theo Adam
        W, V, S = adam_optimizer(W, V, S, dw, lr, beta1, beta2, t)
        
        # Lưu kết quả
        results.append(W.copy())
        
    return results
```

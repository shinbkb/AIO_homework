```python
import cv2
from ultralytics import YOLO
```

```python
model = YOLO("yolov8n.pt")
```

```python
cap = cv2.VideoCapture(1)
print("Hệ thống AI đang hoạt động...")
print("-> BẤM PHÍM 'q' trên cửa sổ camera để DỪNG QUÉT và thoát chương trình.")
```

```python
while cap.isOpened():
    # Đọc từng khung hình (frame) từ camera
    success, frame = cap.read()
    
    if not success:
        print("Không thể kết nối với Camera.")
        break

    # Chạy AI phát hiện người trên frame hiện tại
    # Lệnh model() này trả về kết quả ngay lập tức thay vì chạy lặp vô tận
    results = model(frame, conf=0.45, classes=0, verbose=False)

    # Vẽ khung nhận diện lên ảnh
    annotated_frame = results[0].plot()

    # Hiển thị ảnh đã xử lý lên cửa sổ màn hình
    cv2.imshow("UAV Scanning - ResFly Team", annotated_frame)

    # --- CƠ CHẾ DỪNG QUÉT TẠI ĐÂY ---
    # Chờ 1 mili-giây xem người dùng có bấm phím nào không
    # Nếu phím bấm trùng với ký tự 'q' (Quit), vòng lặp while sẽ bị bẻ gãy (break)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Nhận lệnh dừng quét từ người dùng.")
        break
cap.release()          # Tắt hẳn camera (đèn webcam sẽ tắt)
cv2.destroyAllWindows() # Đóng toàn bộ các cửa sổ hiển thị của OpenCV
print("Đã giải phóng camera thành công.")   
```

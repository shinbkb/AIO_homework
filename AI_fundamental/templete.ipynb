```python
import numpy as np
```

```python
def Euclidean_distance(v1, v2):
    return np.sqrt(sum(abs(a - b)**2 for a, b in zip(v1, v2)))
```

```python
def input_vector(prompt):
    while True:
        try:
            raw_input = input(prompt)
            values = [float(x) for x in raw_input.strip().split()]
            if len(values) != 3:
                print("Vui lòng nhập đúng 3 giá trị đặc trưng cách nhau bởi khoảng trắng!")
                continue
            return values
        except ValueError:
            print("Giá trị không hợp lệ. Vui lòng nhập các số thực!")
```

```python
def main():
    print("=== CHƯƠNG TRÌNH PHÂN LOẠI TÁO VÀ LÊ (2 MẪU MỖI LOẠI) ===\n")
    
    print("--- Nhập dữ liệu cho TÁO ---")
    a1 = input_vector("Nhập đặc trưng TÁO mẫu 1 (A1): ")
    a2 = input_vector("Nhập đặc trưng TÁO mẫu 2 (A2): ")
    
    print("\n--- Nhập dữ liệu cho LÊ ---")
    p1 = input_vector("Nhập đặc trưng LÊ mẫu 1 (P1): ")
    p2 = input_vector("Nhập đặc trưng LÊ mẫu 2 (P2): ")
    
    print("\n--- Nhập dữ liệu TRÁI TEST ---")
    t = input_vector("Nhập đặc trưng Trái test (T): ")
    
    
    d_a1 = Euclidean_distance(t, a1)
    d_a2 = Euclidean_distance(t, a2)
    d_p1 = Euclidean_distance(t, p1)
    d_p2 = Euclidean_distance(t, p2)
    
 
    min_dist_cam = min(d_a1, d_a2)
    min_dist_quyt = min(d_p1, d_p2)
    
    print("\n=== KẾT QUẢ KHOẢNG CÁCH EUCLIDEAN ===")
    print(f"Khoảng cách d(T, A1) [TÁO 1]:  {d_a1:.4f}")
    print(f"Khoảng cách d(T, A2) [TÁO 2]:  {d_a2:.4f}")
    print(f"-> Khoảng cách nhỏ nhất tới TÁO:  {min_dist_cam:.4f}\n")
    
    print(f"Khoảng cách d(T, P1) [LÊ 1]: {d_p1:.4f}")
    print(f"Khoảng cách d(T, P2) [LÊ 2]: {d_p2:.4f}")
    print(f"-> Khoảng cách nhỏ nhất tới LÊ: {min_dist_quyt:.4f}")
    
    print("\n=== KẾT LUẬN ===")
    if min_dist_cam < min_dist_quyt:
        print(f"-> Trái test gần mẫu TÁO nhất (khoảng cách = {min_dist_cam:.4f})")
        print("-> KẾT LUẬN: Trái test là TÁO ")
    elif min_dist_quyt < min_dist_cam:
        print(f"-> Trái test gần mẫu LÊ nhất (khoảng cách = {min_dist_quyt:.4f})")
        print("-> KẾT LUẬN: Trái test là LÊ ")
    else:
        print("-> Trái test có khoảng cách nhỏ nhất bằng nhau tới cả TÁO và LÊ!")
```

```python
if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/Iris.csv")
df.head()
```

```python
def normalization(data, columns):
    result = data.copy()

    for col in columns:
        min_val = result[col].min()
        max_val = result[col].max()

        if max_val != min_val:
            result[col] = (result[col] - min_val) / (max_val - min_val)
        else:
            result[col] = 0

    return result

feature_cols = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

df_normalized = normalization(df, feature_cols)

print(df_normalized.head())
```

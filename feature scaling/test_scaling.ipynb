```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
n_sample = 200
data = {
    'uniform_small':np.random.uniform(0,10,n_sample),
    'gaussian_large':np.random.normal(1000,500,n_sample),
    'outlier_feature':np.random.normal(0,1,n_sample)
}
df=pd.DataFrame(data)

df.loc[0,'outlier_feature']=100
df.loc[1,'outlier_feature']=-50

print('Thống kê dữ liệu gốc')
print(df.describe())

print(df)
```

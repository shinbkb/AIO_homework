```python
import numpy as np
```

```python
def df_w(W):
    w1,w2 = W[0],W[1]
    dw1 = 0.2 * w1
    dw2 = 4*w2
    return np.array([dw1,dw2],dtype=np.float32)

def sgd(W,dw,lr):
    return W-lr * dw
```

```python
def train_p1(optimizer,lr,epochs):
    W = np.array([-5,-2],dtype = np.float32)
    results = [W]
    for i in range(epochs):
        dw = df_w(W)
        W=optimizer(W,dw,lr)
        results.append(W)
    return results
```

```python
learning_rate = 0.04
epochs = 30
history = train_p1(sgd,learning_rate,epochs)
print(f'Điểm minimum sau 30 epoch: {history[-1]}')
```

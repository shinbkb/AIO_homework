```python
import numpy as np
```

```python
def df_w(W):
    w1,w2 = W[0],W[1]
    dw1 = 0.2 * w1
    dw2 = 4*w2
    return np.array([dw1,dw2],dtype=np.float32)
```

```python
def sgd_momentum(W,V,dw,lr,beta):
    V=beta*V+(1-beta)*dw
    W=W-lr*V
    return W,V
```

```python
def train_p1_momentum(lr,beta,epochs):
    W=np.array([-5,-2],dtype=np.float32)
    V=np.array([0,0],dtype=np.float32)
    result=([W])
    for i in range(epochs):
        dw=df_w(W)
        W,V=sgd_momentum(W,V,dw,lr,beta)
        result.append(W)
    return result
    
```

```python
learning_rate = 0.6
epochs = 30
beta = 0.5
history = train_p1_momentum(learning_rate,beta,epochs)
print(f'Điểm minimum sau 30 epoch: {history[-1]}')
```
